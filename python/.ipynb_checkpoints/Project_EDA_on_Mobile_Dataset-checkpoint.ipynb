## Data cleaning

```python
# Change Data Types
df['has_5g'] = df['has_5g'].astype('category')
df['has_nfc'] = df['has_nfc'].astype('category')
df['has_ir_blaster'] = df['has_ir_blaster'].astype('category')
df['processor_manufacturer'] = df['processor_manufacturer'].astype('category')
df['processor_brand'] = df['processor_brand'].astype('category')
df['num_core'] = df['num_core'].astype('category')
df['display_notch_type'] = df['display_notch_type'].astype('category')
df['memory_card_supported'] = df['memory_card_supported'].astype('category')
df['memory_card_slot_type'] = df['memory_card_slot_type'].astype('category')
df['operating_system'] = df['operating_system'].astype('category')
```

```python
display(df.describe(exclude ='object'))
print("*"*200)
display(df.describe(include ='object'))
print("*"*200)
display(df.describe(include ='category'))
```

```python
# rename
df.rename(columns={'rear_cam_1':'primary_rear_camera'},inplace=True)
df.rename(columns={'front_camera_primary':'primary_front_camera'},inplace=True)
```

```python
df.duplicated().sum()
```

```python
# Separate smartphones and feature phones based on their OS family

smartphone_df = df[df['operating_system'].isin(['Android', 'iOS'])]
featurephone_df = df[df['operating_system'].isin(['S30+ / Feature OS', 'RTOS', 'Mocor', 'KaiOS'])]
```

# **Exploratory Data Analysis**

```python
# Make copy for further analysis
sdf = smartphone_df.copy()
fdf = featurephone_df.copy()
```

### Bivariate Analysis

```python
brand_count_df = sdf.groupby('brand_name').count()['model']
top_brand_df = sdf[sdf['brand_name'].isin(brand_count_df[brand_count_df>15].index)].sort_values(by='prices',ascending=False)

plt.figure(figsize=(15,5))
sns.barplot(data = top_brand_df, x = 'brand_name', y = 'prices',palette='winter')
plt.show()
```

```python
sns.scatterplot(data = sdf, x = 'prices', y = 'rating',palette='summer')
```

```python
fig,ax = plt.subplots(nrows=1,ncols=3,figsize=(15,4))

sns.barplot(x = sdf['has_5g'] , y = sdf['prices'],palette='autumn',ax=ax[0])
sns.barplot(x = sdf['has_nfc'] , y = sdf['prices'],palette='Set2',ax=ax[1])
sns.barplot(x = sdf['has_ir_blaster'], y = sdf['prices'],palette='Set2',ax=ax[2])

fig.suptitle("Price vs 5g/nfc/ir_blaster enable")

ax[0].set_xlabel('has_5g')
ax[1].set_xlabel('has_nfc')
ax[2].set_xlabel('has_ir_blaster')

fig.supylabel("Prices")

ax[0].set_ylabel('')
ax[1].set_ylabel('')
ax[2].set_ylabel('')

plt.tight_layout()
plt.show()
```
