```python
with open("../scrapping/final_smartprix.html",'r',encoding='utf-8') as f:
    html = f.read()
```

```python
import pandas as pd
import numpy as np
from bs4 import BeautifulSoup
```

```python
soup = BeautifulSoup(html,'lxml')
```

```python
container = soup.find_all('div',class_='sm-product has-tag has-features has-actions')
```

```python
names = []
prices = []
spec_score = []
sim = []
processor = []
ram = []
battery = []
display = []
camera = []
os = []
card = []
for contain in container:
    try:
        names.append(contain.find('h2').text)
    except:
        names.append(np.nan)

    try:
        prices.append(contain.find('span',class_='price').text)
    except:
        prices.append(np.nan)
        
    try:
        spec_score.append(contain.find('div',class_='score rank-2-bg').find('b').text)
    except:
        spec_score.append(np.nan)

    x = contain.find('ul',class_='sm-feat specs').find_all('li')
    try:
        sim.append(x[0].text)
    except:
        sim.append(np.nan)
    try:
        processor.append(x[1].text)
    except:
        processor.append(np.nan)
    try:    
        ram.append(x[2].text)
    except:
        ram.append(np.nan)
    try:
        battery.append(x[3].text)
    except:
        battery.append(np.nan)
    try:
        display.append(x[4].text)
    except:
        display.append(np.nan)
    try:
        camera.append(x[5].text)
    except:
        camera.append(np.nan)
    try:
        card.append(x[6].text)
    except:
        card.append(np.nan)
    try:
        os.append(x[7].text)
    except:
        os.append(np.nan)
```

```python
len(display)
```

```python
df = pd.DataFrame({
        'model':names,
        'prices':prices,
        'rating':spec_score,
        'sim':sim,
        'processor':processor,
        'ram':ram,
        'battery':battery,
        'display':display,
        'camera':camera,
        'card':card,
        'os':os
        })
```

```python
df
```

```python
df.to_csv("../dataset/Smartphone_uncleaned.csv",index=False)
```
